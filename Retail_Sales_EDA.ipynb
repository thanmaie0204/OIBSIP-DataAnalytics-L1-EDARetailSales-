```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Sales", bins=30, kde=True)

plt.title("Distribution of Sales")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="Category", y="Sales", estimator=sum, errorbar=None)

plt.title("Total Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="Region", y="Sales", estimator=sum, errorbar=None)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```

```python
# Find the top 10 products by sales
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

print(top_products)
```

```python
plt.figure(figsize=(12, 6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Total Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")

plt.tight_layout()
plt.show()
```

```python
profit_category = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

print(profit_category)
```

```python
profit_category.plot(kind="bar", figsize=(8, 5))

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)

plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()

monthly_sales.plot(figsize=(12, 5), marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)

plt.show()
```

```python
# Total sales by customer segment
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

print(segment_sales)
```

```python
segment_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Total Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()
```

```python
# Select numerical columns
numerical_data = df.select_dtypes(include="number")

# Create correlation matrix
correlation = numerical_data.corr()

# Plot heatmap
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

```python
# Calculate mean, median, mode, and standard deviation
sales_statistics = {
    "Mean": df["Sales"].mean(),
    "Median": df["Sales"].median(),
    "Mode": df["Sales"].mode()[0],
    "Standard Deviation": df["Sales"].std()
}

print("Sales Statistics:")
for statistic, value in sales_statistics.items():
    print(f"{statistic}: {value:.2f}")
```

```python
# Extract year and quarter
df["Year"] = df["Order Date"].dt.year
df["Quarter"] = df["Order Date"].dt.to_period("Q")

# Calculate quarterly sales
quarterly_sales = df.groupby("Quarter")["Sales"].sum()

print(quarterly_sales)
```

```python
quarterly_sales.plot(figsize=(12, 5), marker="o")

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```
